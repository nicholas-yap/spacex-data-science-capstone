# Interactive launch-site maps

Folium maps use the 90-row dataset and OpenStreetMap tiles. The coordinates represent launch sites. MarkerCluster spreads overlapping markers only for display. Proximity distances refer to selected sample points, not surveyed minimum distances.

```python
from pathlib import Path
import os
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
os.chdir(root)
assert (root / 'data').is_dir(), 'Run from the repository root or notebooks directory.'

import runpy
runpy.run_path('build_maps.py')
print('Created maps/launch_sites.html, maps/launch_outcomes.html and maps/proximity.html')
```

```python
from IPython.display import IFrame, display
for name in ['launch_sites','launch_outcomes','proximity']:
    display(IFrame(src='../maps/'+name+'.html',width=1000,height=600))
```
