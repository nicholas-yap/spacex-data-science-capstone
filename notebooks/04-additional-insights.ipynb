# Additional insights

This analysis supports the presentation’s additional insights using the saved 90-row course dataset. Results describe historical associations. They do not isolate causal effects or estimate launch price.

```python
from pathlib import Path
import os
import pandas as pd
root=Path.cwd()
if root.name=='notebooks':root=root.parent
os.chdir(root)
df=pd.read_csv('data/dataset_part_2.csv')
assert len(df)==90
assert int(df.Class.sum())==60
print('90 launches and 60 Class 1 outcomes')
```

## Launch count and success rate answer different questions
A site can contribute the largest success count while a smaller site has a higher observed success rate.

```python
site=df.groupby('LaunchSite').Class.agg(Launches='count',Successes='sum',Rate='mean')
display(site)
assert int(site.loc['CCAFS SLC 40','Successes'])==33
assert int(site.loc['KSC LC 39A','Successes'])==17
print('CCAFS leads success count. KSC leads observed rate.')
```

## The course label includes controlled ocean outcomes
ASDS and RTLS successes are distinct from the five True Ocean outcomes. Class 1 should not directly substitute for a count of booster landings on a drone ship or ground pad.

```python
ocean=df.Outcome.eq('True Ocean').sum()
pad=df.Outcome.isin(['True ASDS','True RTLS']).sum()
assert int(ocean)==5 and int(pad)==55
results=pd.DataFrame({'Measure':['Course Class 1','Successful ASDS or RTLS','Controlled ocean outcomes'],'Count':[60,int(pad),int(ocean)],'Share of 90 launches':[60/90,pad/90,ocean/90]})
display(results)
print(f'Including controlled ocean outcomes raises the Class 1 rate by {ocean/90*100:.1f} percentage points.')
```

## Booster blocks and sample size
Compare both counts and rates. Block 5 has a larger sample than Block 2. Date and technology overlap, so group differences alone cannot identify a causal effect.

```python
block=df.groupby('Block').Class.agg(Launches='count',Successes='sum',Rate='mean')
display(block)
assert int(block.loc[5,'Launches'])==39 and int(block.loc[5,'Successes'])==33
assert int(block.loc[1,'Launches'])==19 and int(block.loc[1,'Successes'])==4
assert int(block.loc[2,'Launches'])==6 and int(block.loc[2,'Successes'])==6
print('Block 5: 33/39 = 84.6%. Block 1: 4/19 = 21.1%. Block 2: 6/6, with only six observations.')
```
